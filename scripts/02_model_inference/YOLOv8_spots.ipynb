```python
from ultralytics import YOLO


def run_yolov8_segmentation(
    model_path="../../model_weights/yolov8_spots.pt",
    source_path="../../data/stressed/images/operculum_images",
    conf=0.25,
    iou=0.15,
    device="cpu"
):
    """
    Run YOLOv8 segmentation inference.

    Use device='cpu' if torchvision CUDA NMS is not working.
    Use device=0 or device='cuda' after fixing PyTorch/torchvision CUDA.
    """

    model = YOLO(model_path)

    results = model.predict(
        source=source_path,
        conf=conf,
        iou=iou,
        device=device,
        save_txt=True,
        save_conf=True,
        save=True,
        show_conf=True,
        show_labels=True,
        retina_masks=True,
        show_boxes=True
    )

    return results
```

```python
results = run_yolov8_segmentation(
    model_path="../../model_weights/yolov8_spots.pt",
    source_path="../../data/control/images/operculum_images",
    device="gpu"
)
```
